```python
A = np.array([[2., 1.], [1., 3.]])

A.T # transpose

np.linalg.inv(A) # inverse

np.trace(A) # sum of diagonal = 5

np.allclose(A, A.T) # symmetric?

```

```python
theta = np.radians(25)
point=np.sin(theta), np.cos(theta)


R = np.array([[np.cos(theta), -np.sin(theta)],

[np.sin(theta), np.cos(theta)]])

R @ point # rotate a point about the origin
```
