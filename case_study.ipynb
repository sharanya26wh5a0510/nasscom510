```python
time = [0, 1, 2, 3, 4, 5]
temperature = [32, 35, 38, 40, 42, 45]
voltage = [400, 395, 390, 385, 380, 375]
current = [20, 22, 25, 28, 30, 32]
charge = [95, 88, 80, 70, 60, 48]
```

```python
for i in range(6):
    print(time[i], temperature[i], voltage[i], current[i], charge[i])
```

```python
for i in range(6):
    if temperature[i] > 40:
        print("High Temperature Alert")
    else:
        print("Temperature Normal")
```

```python
for i in range(6):
    if charge[i] < 50:
        print("Low Battery Alert")
    else:
        print("Battery Charge Normal")
```

```python
def battery_status(c):
    if c > 70:
        return "High"
    elif c >= 50:
        return "Medium"
    else:
        return "Low"

print(battery_status(95))
print(battery_status(60))
print(battery_status(48))
```

```python
import numpy as np

D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])
```

```python
print("Shape:", D.shape)
```

```python
print("Average temperature:", np.mean(D[:, 0]))
print("Average voltage:", np.mean(D[:, 1]))
print("Average current:", np.mean(D[:, 2]))
print("Average charge:", np.mean(D[:, 3]))
```

```python
print("Maximum temperature:", np.max(D[:, 0]))
print("Minimum temperature:", np.min(D[:, 0]))
```

```python
voltage = D[:, 1]
print("Voltage:", voltage)
```

```python
print(D[D[:, 0] > 40])
```

```python
print(D[D[:, 3] < 50])
```

```python
A = np.array([[2, 1],
              [1, 3]])

X = np.array([40, 25])

Y = A @ X

print("Y:", Y)
```

```python
print("Transpose:")
print(A.T)
```

```python
print("Determinant:", np.linalg.det(A))
```

```python
print("Inverse:")
print(np.linalg.inv(A))
```

```python
A = np.array([[2, 1],
              [1, 3]])

B = np.array([100, 120])

answer = np.linalg.solve(A, B)

print("x and y:", answer)
```
