```python
import numpy as np
import sympy as sp

x, y =sp.symbols('x y')
sp.init_printing()
np.random.seed(52)
print('setup complete. sympy',sp.__version__,'|numpy',np.__version__)
```

```python


f2 = x**3 + 6*x*y + y**9
print('df/dx =', sp.diff(f2, x))     # 2*x + 3*y
print('df/dy =', sp.diff(f2, y))     # 3*x + 2*y
```

```python
# 1. dh/dx and dh/dy
dh_dx = sp.diff(f2, x)
dh_dy = sp.diff(f2, y)

# 2. Assemble the gradient list
gradient = [dh_dx, dh_dy]

# 3. Evaluate at (x=2, y=0)
gradient_at_point = [g.subs({x: 2, y: 0}) for g in gradient]
```

```python



X = np.random.randn(6, 4)
Y = np.random.randn(6, 1)
W1 = np.random.randn(4, 6) * 0.2
W2 = np.random.randn(6, 1) * 0.2

z1 = X @ W1
h = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()

print('Initial loss:', round(loss, 4))
```

```python



dy  = 2 * (y_hat - Y) / Y.size
dW2 = h.T @ dy
dh  = dy @ W2.T
dz1 = dh * (z1 > 0)
dW1 = X.T @ dz1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python


lr = 0.1

W1 -= lr * dW1
W2 -= lr * dW2

h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()

print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
#Lab Excersise
#Forward
z1 = X @ W1
h = np.maximum(0, z1)
y_hat = h @ W2
loss_before = ((y_hat - Y) ** 2).mean()

# Backward
dy = 2 * (y_hat - Y) / Y.size
dW2 = h.T @ dy
dh = dy @ W2.T
dz1 = dh * (z1 > 0)
dW1 = X.T @ dz1

# Update
lr = 0.05
W1 -= lr * dW1
W2 -= lr * dW2

# Loss after update
y_hat = np.maximum(0, X @ W1) @ W2
loss_after = ((y_hat - Y) ** 2).mean()

print("Loss before:", round(loss_before, 4))
print("Loss after:", round(loss_after, 4))
```
