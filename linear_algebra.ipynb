```python
LAB EXCERSISE-3

A = np.array([[2., 1.],
              [1., 3.]])

B = np.array([[1., 0.],
              [4., 2.]])

# Matrix multiplication
print('A @ B:\n', A @ B)

# Transpose
print('A.T:\n', A.T)

# Inverse
print('inverse(A):\n', la.inv(A))

# Trace
print('trace(A) = sum of diagonal:', np.trace(A))

# Identity matrix
I = np.eye(2)
print('Identity I:\n', I)

# Verify inverse
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))

# Check symmetric
print('A symmetric?', np.allclose(A, A.T))

# Orthogonal matrix Q
theta = np.radians(30)

Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
M = np.array([[4., 2.],
              [2., 3.]])

P = np.array([[0., -1.],
              [1.,  0.]])

# 1. inverse and trace of M
print('inverse(M):\n', la.inv(M))
print('trace(M):', np.trace(M))

# 2. M @ inv(M) == I ?
I = np.eye(2)
print('M @ inv(M) == I ?', np.allclose(M @ la.inv(M), I))

# 3. Is M symmetric? Is P orthogonal?
print('M symmetric?', np.allclose(M, M.T))
print('P orthogonal?', np.allclose(P.T @ P, I))
```

```python

square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)


S = np.array([[1.5, 0.0],
              [0.0, 0.5]])


t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python

# LAB EXERCISE 4

# Triangle: each column is a point
tri = np.array([[0, 1, 0],
                [0, 0, 1]], dtype=float)

# 1. Scaling matrix
# Double x and halve y
S = np.array([[2., 0.],
              [0., 0.5]])

# 2. Rotation matrix for 45 degrees
t = np.radians(45)

R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

# 3. Apply both transformations one after another
scaled = S @ tri
transformed = R @ scaled

# Print transformed triangle
print('Transformed triangle:\n', transformed)

# Function to close the triangle
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

# Plot original and transformed triangle
fig, ax = plt.subplots(figsize=(5, 5))

ax.plot(*close_loop(tri), marker='o', label='original')
ax.plot(*close_loop(transformed), marker='o',
        label='scaled + rotated 45 deg')

ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)

ax.set_aspect('equal')
ax.legend()
ax.set_title('Scaling and Rotation')
plt.show()
```

```python
tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])   # 3 corners of a triangle

# 1. Scaling matrix (x*2, y*0.5)
S = np.array([[2.0, 0.0],
              [0.0, 0.5]])

# 2. Rotation matrix (45 degrees)
t = np.radians(45)

R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

# 3. Apply both transformations
scaled = S @ tri
transformed = R @ scaled

# Close the triangle
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])

# Plot original vs transformed
fig, ax = plt.subplots(figsize=(5, 5))

ax.plot(*close_loop(tri), marker='o', label='original')
ax.plot(*close_loop(transformed), marker='o',
        label='scaled + rotated 45 deg')

ax.axhline(0, color='gray', lw=0.5)
ax.axvline(0, color='gray', lw=0.5)

ax.set_aspect('equal')
ax.legend()
ax.set_title('Scaling and Rotation')
plt.show()
```

```python

A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python

for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
C = np.array([[4., 1.],
              [2., 3.]])

# 1. Eigenvalues and eigenvectors
vals, vecs = la.eig(C)

# 2. Print eigenvalues
print('Eigenvalues:', vals)

# 3. Verify C v = lambda v for the first eigenvector
v = vecs[:, 0]

print('First eigenvector:', v)

print('C @ v == lambda * v ?',
      np.allclose(C @ v, vals[0] * v))
```
